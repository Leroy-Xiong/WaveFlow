# WaveFlow figure reproduction

This notebook renders the released paper panels one at a time. Each cell runs
one release command and displays its output immediately.

The rendering path needs the paper data tree at the repository root. The
case forcing (ERA5 and IFS ENS slices) is downloaded by the setup cell.

```python
import os, shutil, subprocess, sys
from pathlib import Path
from IPython.display import Image, display

if not Path('reproduce.py').exists():
    os.chdir('..')
PYTHON = os.environ.get('WAVEFLOW_PYTHON')
if not PYTHON and os.environ.get('CONDA_PREFIX'):
    PYTHON = str(Path(os.environ['CONDA_PREFIX']) / 'bin' / 'python')
PYTHON = PYTHON or shutil.which('python') or sys.executable
print('working directory:', Path.cwd())
print('python:', PYTHON)
subprocess.run([PYTHON, 'reproduce/download.py', '--forcing'], check=True)
sys.path.insert(0, str(Path('reproduce').resolve()))
import notebook_render
```
