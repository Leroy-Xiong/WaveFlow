# WaveFlow figure reproduction

This notebook downloads the case forcing and renders the released paper
panels (Fig. 2a, 3d-f, 5g-i, 6, ED 2, 3, 5, 6). The rendering path needs the
paper data tree; without it, the notebook falls back to the public inference
example. An NVIDIA GPU is required for `infer`.

## Rendered panels

![All released panels](preview_all.png)

The executed run below renders all 13 panels; the full-size PNGs are written
to `reproduced/` when the notebook is run locally.

```python
import os
import shutil
import subprocess
import sys
from pathlib import Path
from IPython.display import Image, display

if not Path('reproduce.py').exists():
    os.chdir('..')
PYTHON = os.environ.get('WAVEFLOW_PYTHON')
if not PYTHON and os.environ.get('CONDA_PREFIX'):
    PYTHON = str(Path(os.environ['CONDA_PREFIX']) / 'bin' / 'python')
PYTHON = PYTHON or shutil.which('python') or sys.executable
print('working directory:', Path.cwd())
print('python:', PYTHON)
```

```python
subprocess.run([PYTHON, 'reproduce/download.py', '--forcing'], check=True)
```

## 1. Render all released panels

The `all` command renders every panel that has its data available and skips
the ones that need a third-party hindcast frame. `verify` compares the result
against `reference/`.

```python
if Path('data/hk_coastline').exists():
    subprocess.run([PYTHON, 'reproduce.py', 'all'], check=True)
    subprocess.run([PYTHON, 'reproduce.py', 'verify'], check=True)
else:
    print('paper data tree not found; running the public Fig. 3f example')
    subprocess.run([PYTHON, 'reproduce.py', 'infer', 'fig3f'], check=True)
```

```python
for p in sorted(Path('reproduced').glob('*.png')):
    display(Image(str(p)))
```
