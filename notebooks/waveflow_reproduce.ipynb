# WaveFlow figure reproduction

This notebook renders each paper panel as an independent figure, one at a
time. The panel functions live in `reproduce/notebook_standalone.py`; they do
not crop the original paper canvas.

Fig. 2a runs from the public model assets. The remaining panels need the paper
data tree at the repository root.

```python
import os, shutil, subprocess, sys
from pathlib import Path
from IPython.display import Image, display

if not Path('reproduce.py').exists():
    os.chdir('..')
sys.path.insert(0, str(Path('reproduce').resolve()))
import notebook_render, notebook_standalone
PYTHON = os.environ.get('WAVEFLOW_PYTHON')
if not PYTHON and os.environ.get('CONDA_PREFIX'):
    PYTHON = str(Path(os.environ['CONDA_PREFIX']) / 'bin' / 'python')
PYTHON = PYTHON or shutil.which('python') or sys.executable
subprocess.run([PYTHON, 'reproduce/download.py', '--forcing'], check=True)
```
